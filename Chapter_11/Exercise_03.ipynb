```python
def most_frequent_letters (cadena):
    diccionario = {}
    contador = 0
    for value in cadena:
        if value in diccionario:
            diccionario[value] += 1
        else:
            diccionario[value] = 1
    return sorted(diccionario.items(), key=second_Element, reverse=True)


def second_Element (t):
    return t[1]
```
